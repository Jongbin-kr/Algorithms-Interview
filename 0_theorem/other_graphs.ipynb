### 서로소 집합 알고리즘
- $O(V + M(1 + \log_{2-M/V}V)) \simeq O(V + M\log_2 V)$ 

```python
def find_parent(parent_table, x):
    print(f"x: {x}, parent_table[x]: {parent_table[x]}")
    if parent_table[x] != x:
        parent_table[x] = find_parent(parent_table, parent_table[x])
    return x

def union_parent(parent_table, a, b):
    a = find_parent(parent_table, a)
    b = find_parent(parent_table, b)
    print(a, b)
    ## 더 작은 값을 부모로 설정
    if a < b: parent_table[b] = a
    else: parent_table[a] = b
    
    
    
v, e = map(int, input().split())
parent_table = [0] * (v+1)
for i in range(1, v+1):     ## 부모를 자기 자신으로 초기화
    parent_table[i] = i
    
for _ in range(e):
    a, b = map(int, input().split())
    union_parent(parent_table, a, b)
    print(parent_table, end='\n\n')
    


## 각 원소가 속한 집합 출력
for i in range(1, v+1):
    print(find_parent(parent_table, i))
    
print(parent_table)
```

### 최소 신장 트리 (kruskal) 알고리즘

모든 정점을 연결하면서 간선의 비용 합을 최소화하는 게 목표야. 불필요한 순환 없이 연결하면 다리는 n−1개가 되고, 그 구조를 최소 신장 트리라고 해.

```python
# programmers/high-score-kit/6_greedy/섬 연결하기.py
def solution(n, costs):
    costs.sort(key=lambda x: x[2])
    group = list(range(n))
    sum_costs = 0
    
    for start, end, cost in costs:
        start_group = group[start]
        end_group = group[end]
        
        if group[start] == group[end]: continue
        
        for island in range(n):
            if group[island] == end_group:
                group[island] = start_group
                
        sum_costs += cost
            
    return sum_costs
    
```
